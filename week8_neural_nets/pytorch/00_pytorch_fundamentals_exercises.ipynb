```python
torch.cuda.manual_seed_all(1234)
tensor_A = torch.rand(size=(2, 3)).to("cuda")
tensor_B = torch.rand(size=(2, 3)).to("cuda")
print(f"Tensor A:\n{tensor_A}\nDevice: {tensor_A.device}\n")
print(f"Tensor B:\n{tensor_B}\nDevice: {tensor_B.device}")
```

```python
res = torch.matmul(tensor_A, tensor_B.T)
res
```

```python
torch.max(res), torch.min(res), torch.argmax(res), torch.argmin(res)
```
