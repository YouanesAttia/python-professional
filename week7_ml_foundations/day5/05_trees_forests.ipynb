```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.tree import plot_tree
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt


iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names
class_names = iris.target_names

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)

tree = DecisionTreeClassifier(random_state=42)
tree.fit(X_train, y_train)
plot_tree(tree, feature_names=feature_names, class_names=class_names, filled=True, rounded=True)
plt.title("Decision Tree Visualization (Full Depth)")
plt.show()
```

```python
tree_unconstrained = DecisionTreeClassifier(max_depth=None, random_state=42)
tree_unconstrained.fit(X_train, y_train)

tree_constrained = DecisionTreeClassifier(max_depth=3, random_state=42)
tree_constrained.fit(X_train, y_train)

print(f'(depth=None) - Train acc: {tree_unconstrained.score(X_train, y_train)}')
print(f'(depth=None) - Test acc: {tree_unconstrained.score(X_test, y_test)}')

print(f'(depth=3) - Train acc: {tree_constrained.score(X_train, y_train)}')
print(f'(depth=3) - Test acc: {tree_constrained.score(X_test, y_test)}')
```
