```python
import pandas as pd
import numpy as np

data = pd.read_csv('examples/2015.csv')
data.head()
```

```python
data.groupby('Region')['Happiness Score'].mean()
```

```python
data.groupby('Region').agg({'Happiness Score': 'mean',
                            'Economy (GDP per Capita)': ['min', 'max']})
```

```python
z_score_fn = lambda x: (x - x.mean()) / x.std()
data['Regional_Happiness_ZScore'] = data.groupby('Region')['Happiness Score'].transform(z_score_fn)
data.head()
```

```python
df_scores = data[['Country', 'Happiness Score']].head(5)
df_regions = data[['Country', 'Region']].iloc[2:7]
inner = pd.merge(df_scores, df_regions, how='inner')
outer = pd.merge(df_scores, df_regions, how='outer')
right = pd.merge(df_scores, df_regions, how='right')
left = pd.merge(df_scores, df_regions, how='left')
display(outer)
```

```python
top_3 = data.head(3)
bottom_3 = data.tail(3)
vertical = pd.concat([top_3, bottom_3])

country = data[['Country']].head()
Zscore = data[['Regional_Happiness_ZScore']].head()
horizontal = pd.concat([country, Zscore], axis=1)

display(horizontal)
display(vertical)
```
